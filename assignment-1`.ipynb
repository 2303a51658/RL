```python
# Multi-Armed Bandit - Greedy Algorithm
# Part 1: Arithmetic Mean
# Part 2: Incremental Update


def arithmetic_mean_greedy(machines, rewards):
    print("\n===== PART 1: ARITHMETIC MEAN METHOD =====")

    q_values = {}

    for machine in machines:
        reward_list = rewards[machine]

        # Calculate arithmetic mean
        q = sum(reward_list) / len(reward_list)

        q_values[machine] = q

        print(f"\nMachine {machine}")
        print(f"Rewards: {reward_list}")
        print(f"Q({machine}) = {q:.2f}")

    # Greedy selection
    selected_machine = max(q_values, key=q_values.get)

    print(f"\nGreedy Selected Machine: {selected_machine}")


def incremental_greedy(machines, rewards):
    print("\n===== PART 2: INCREMENTAL UPDATE METHOD =====")

    # Initialize Q and N
    q_values = {machine: 0.0 for machine in machines}
    counts = {machine: 0 for machine in machines}

    print("\nInitial Values:")
    for machine in machines:
        print(f"Machine {machine}: Q = {q_values[machine]}, N = {counts[machine]}")

    print("\nUpdates after every play:")

    # Process every observed reward
    for machine in machines:
        for reward in rewards[machine]:

            # Increment number of plays
            counts[machine] += 1

            # Incremental update formula
            q_values[machine] = (
                q_values[machine]
                + (reward - q_values[machine]) / counts[machine]
            )

            # Find current greedy machine
            selected_machine = max(q_values, key=q_values.get)

            print(
                f"Machine {machine}, Reward = {reward}, "
                f"N({machine}) = {counts[machine]}, "
                f"Q({machine}) = {q_values[machine]:.2f}, "
                f"Greedy Selection = {selected_machine}"
            )

    print("\nFinal Estimated Values:")

    for machine in machines:
        print(
            f"Machine {machine}: "
            f"Q({machine}) = {q_values[machine]:.2f}, "
            f"N({machine}) = {counts[machine]}"
        )

    # Final Greedy selection
    selected_machine = max(q_values, key=q_values.get)

    print(f"\nFinal Greedy Selected Machine: {selected_machine}")


# ---------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------

# Take number of machines as input
n = int(input("Enter the number of machines: "))

machines = []
rewards = {}

# Take machine names and rewards as input
for i in range(n):
    machine = input(f"\nEnter name of machine {i + 1}: ")
    machines.append(machine)

    reward_input = input(
        f"Enter observed rewards for Machine {machine} "
        "(separated by spaces): "
    )

    reward_list = list(map(float, reward_input.split()))
    rewards[machine] = reward_list


# Run arithmetic mean method
arithmetic_mean_greedy(machines, rewards)

# Run incremental update method
incremental_greedy(machines, rewards)
```
